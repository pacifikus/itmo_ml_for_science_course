#### ndarray

Массив — упорядоченная изменяемая структура данных, к элементам которой можно обратиться по индексу

Основным объектом NumPy является **ndarray** (n-dimensional array) - структура данных, которая позволяет хранить набор элементов одного типа

Массивы NumPy имеют фиксированный размер при создании, в отличие от списков Python (которые могут динамически увеличиваться). Изменение размера ndarray создаст новый массив и удалит исходный.

### Почему форма и операции важны?

```python
prediction = np.array([0.0, 1.0, 0.0])
target = np.array([[0.0], [1.0], [0.0]])

prediction, target
```

```python
wrong_errors = prediction - target
wrong_mse = np.mean(wrong_errors ** 2)
```

```python
prediction.shape, target.shape, wrong_errors.shape
```

```python
wrong_errors
```

```python
wrong_mse
```

image.png

numpy сопоставляет размеры массивов справа налево. У prediction форма (3,), а у target - (3, 1). Для сопоставления NumPy мысленно дополняет форму prediction слева единицей

```python
correct_target = target[:, 0]
correct_errors = prediction - correct_target
correct_mse = np.mean(correct_errors ** 2)
```

```python
correct_errors.shape, correct_mse
```

Но не все так однозначно...

```python
source_values = np.array([10, 20, 30, 40, 50])
sliced_values = source_values[1:4]
selected_values = source_values[[1, 3]]
independent_values = source_values[1:4].copy()

print("Срез разделяет память:", np.shares_memory(source_values, sliced_values))
print("Выбор индексов разделяет память:", np.shares_memory(source_values, selected_values))
print("Явная копия разделяет память:", np.shares_memory(source_values, independent_values))
```

```python
sliced_values[0] = 999
independent_values[1] = -1

source_values, independent_values
```

Общее правило - базовая индексация возвращает представление массива, а расширенная индексация при чтении возвращает копию

И еще неожиданнности...

```python
values = np.arange(10)

even_values = values[::2]
odd_values = values[1::2]
first_half = values[:5]
```

```python
even_values, odd_values, first_half
```

```python
print(np.shares_memory(values, even_values))
print(np.shares_memory(even_values, odd_values))
print(np.shares_memory(values, first_half))
```

С reshape тоже не все так просто...

```python
matrix = np.arange(12, dtype=np.int32).reshape(3, 4)
matrix
```

```python
left_columns = matrix[:, :2]
left_columns_flat = left_columns.reshape(-1)

even_columns = matrix[:, ::2]
even_columns_flat = even_columns.reshape(-1)
```

```python
left_columns_flat, even_columns_flat
```

```python
print("Два столбца:\n", left_columns)
print("Их strides:", left_columns.strides)
print("Получился view:", np.shares_memory(matrix, left_columns_flat))
```

```python
print("Каждый второй столбец:\n", even_columns)
print("Их strides:", even_columns.strides)
print("Получился view:", np.shares_memory(matrix, even_columns_flat))
```

Для представления без копии NumPy требуется подобрать новые shape и strides, которые адресуют те же элементы исходного буфера в нужном порядке. Если между последовательными элементами результата потребовались бы разные, невыразимые одним шагом переходы, данные копируются

### Проверка на равенство и близость

`==` для результатов вычислений с плавающей точкой часто слишком строгое

`np.allclose` сравнивает элементы с допущением

для поиска `NaN` и бесконечностей используйте `np.isfinite`.

```python
decimal_values = np.array([0.1, 0.2, 0.3])
computed_sum = decimal_values.sum()
expected_sum = 0.6

computed_sum == expected_sum
```

```python
np.isclose(computed_sum, expected_sum, rtol=1e-12, atol=1e-12)
```
